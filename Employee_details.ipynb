```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# ---------------------------
# 1. Load Dataset
# ---------------------------
df = pd.read_csv("Employee-Attrition - Employee-Attrition.csv")

# ---------------------------
# 2. Remove Unwanted Columns
# ---------------------------
# These columns add no predictive value or are unique identifiers
drop_cols = [
    "EmployeeCount", "EmployeeNumber", "Over18", "StandardHours"
]

df.drop(columns=drop_cols, inplace=True, errors="ignore")

# ---------------------------
# 3. Handle Missing / NaN Values
# ---------------------------
# Numerical → fill with median
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Categorical → fill with mode
cat_cols = df.select_dtypes(include=['object']).columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])

# ---------------------------
# 4. Handle Outliers (IQR Method)
# ---------------------------
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    # Cap outliers (winsorization)
    df[col] = np.where(df[col] < lower, lower,
                       np.where(df[col] > upper, upper, df[col]))

# ---------------------------
# 5. Encode Categorical Variables
# ---------------------------
le = LabelEncoder()

for col in cat_cols:
    df[col] = le.fit_transform(df[col])

# ---------------------------
# 6. Save the Cleaned CSV
# ---------------------------
df.to_csv("employee_attrition_preprocessed.csv", index=False)

print("Preprocessing Completed Successfully!")
print("Saved File: employee_attrition_preprocessed.csv")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

# ---------------------------------------------
# 1. LOAD DATA (preprocessed file)
# ---------------------------------------------
df = pd.read_csv("employee_attrition_preprocessed.csv")

# ---------------------------------------------
# 2. DEFINE TARGET & FEATURES
# ---------------------------------------------
target = "Attrition"

# BEST predictive features
important_features = [
    "OverTime",
    "MonthlyIncome",
    "JobSatisfaction",
    "YearsAtCompany",
    "TotalWorkingYears",
    "JobLevel",
    "Age",
    "EnvironmentSatisfaction",
    "JobRole"
]

X = df[important_features]
y = df[target]   # DO NOT DROP ATTRITION

# ---------------------------------------------
# 3. TRAIN / TEST SPLIT
# ---------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# ---------------------------------------------
# 4. SCALING
# ---------------------------------------------
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled  = scaler.transform(X_test)

# ---------------------------------------------
# 5. LOGISTIC REGRESSION
# ---------------------------------------------
log_reg = LogisticRegression(class_weight="balanced", max_iter=2000)
log_reg.fit(X_train_scaled, y_train)

# ---------------------------------------------
# 6. PREDICT
# ---------------------------------------------
y_pred = log_reg.predict(X_test_scaled)
y_proba = log_reg.predict_proba(X_test_scaled)[:, 1]

# ---------------------------------------------
# 7. METRICS
# ---------------------------------------------
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)
cm = confusion_matrix(y_test, y_pred)

print("\n========== LOGISTIC REGRESSION RESULTS ==========\n")
print(f"Accuracy       : {acc:.4f}")
print(f"Precision      : {prec:.4f}")
print(f"Recall         : {rec:.4f}")
print(f"F1 Score       : {f1:.4f}")
print(f"AUC-ROC Score  : {auc:.4f}")

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, digits=4))

# ---------------------------------------------
# 8. ROC CURVE
# ---------------------------------------------
fpr, tpr, _ = roc_curve(y_test, y_proba)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0,1], [0,1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Logistic Regression (Best Features)")
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

# -----------------------------
# 1. LOAD DATA (preprocessed file)
# -----------------------------
df = pd.read_csv("employee_attrition_preprocessed.csv")

# -----------------------------
# 2. DEFINE TARGET & FEATURES
# -----------------------------
target = "Attrition"

important_features = [
    "OverTime",
    "MonthlyIncome",
    "JobSatisfaction",
    "YearsAtCompany",
    "TotalWorkingYears",
    "JobLevel",
    "Age",
    "EnvironmentSatisfaction",
    "JobRole"
]

X = df[important_features].copy()
y = df[target]   # DO NOT DROP ATTRITION

# -----------------------------
# 3. TRAIN / TEST SPLIT
# -----------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

# -----------------------------
# 4. PREPROCESSING (One-Hot for categorical features)
# -----------------------------
# Detect categorical columns among the chosen features
cat_cols = [c for c in X_train.columns if X_train[c].dtype == "object" 
            or X_train[c].dtype.name == "category"]
num_cols = [c for c in X_train.columns if c not in cat_cols]

print("Numerical cols:", num_cols)
print("Categorical cols:", cat_cols)

# If there are categorical columns, set up ColumnTransformer with OneHotEncoder.
# Use sparse_output=False for newer sklearn versions.
if len(cat_cols) > 0:
    preprocessor = ColumnTransformer([
        ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols)
    ], remainder="passthrough")  # passthrough numeric columns
else:
    # No categorical columns — passthrough all features
    preprocessor = "passthrough"

# -----------------------------
# 5. DECISION TREE PIPELINE
# -----------------------------
dtree = DecisionTreeClassifier(class_weight="balanced", random_state=42)

pipe = Pipeline([
    ("pre", preprocessor),
    ("clf", dtree)
])

# Fit
pipe.fit(X_train, y_train)

# -----------------------------
# 6. PREDICTIONS
# -----------------------------
y_pred = pipe.predict(X_test)
# DecisionTree supports predict_proba
y_proba = pipe.predict_proba(X_test)[:, 1]

# -----------------------------
# 7. METRICS
# -----------------------------
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)
cm = confusion_matrix(y_test, y_pred)

print("\n========== DECISION TREE RESULTS ==========\n")
print(f"Accuracy       : {acc:.4f}")
print(f"Precision      : {prec:.4f}")
print(f"Recall         : {rec:.4f}")
print(f"F1 Score       : {f1:.4f}")
print(f"AUC-ROC Score  : {auc:.4f}")

print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(classification_report(y_test, y_pred, digits=4))

# -----------------------------
# 8. ROC CURVE
# -----------------------------
fpr, tpr, _ = roc_curve(y_test, y_proba)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0,1], [0,1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Decision Tree")
plt.legend()
plt.show()

# -----------------------------
# 9. FEATURE IMPORTANCES (map back to names)
# -----------------------------
# Build feature name list consistent with preprocessor output
feature_names = []
if len(cat_cols) > 0:
    ohe = pipe.named_steps["pre"].named_transformers_["ohe"]
    if hasattr(ohe, "get_feature_names_out"):
        ohe_feature_names = list(ohe.get_feature_names_out(cat_cols))
    else:
        # fallback (older sklearn) - attempt manual expansion (may be empty)
        ohe_feature_names = []
    # remainder passthrough numeric columns come after OHE columns
    feature_names = ohe_feature_names + num_cols
else:
    # passthrough: original column order preserved
    feature_names = num_cols  # which is same as all chosen features

# Get importances from the trained tree
importances = pipe.named_steps["clf"].feature_importances_
feat_imp = pd.Series(importances, index=feature_names).sort_values(ascending=False)

# Plot top 20 if available
top_n = min(20, feat_imp.shape[0])
plt.figure(figsize=(8,6))
feat_imp.head(top_n).plot.barh()
plt.gca().invert_yaxis()
plt.title("Top Feature Importances (Decision Tree)")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

# -----------------------------
# 10. Save test predictions (optional)
# -----------------------------
test_pred_df = X_test.copy()
test_pred_df["y_true"] = y_test.values
test_pred_df["y_pred_dt"] = y_pred
test_pred_df["y_proba_dt"] = y_proba
test_pred_df.to_csv("test_predictions_decision_tree.csv", index=False)
print("Saved test predictions to test_predictions_decision_tree.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

# ---------------------------------------------
# 1. LOAD PREPROCESSED DATA
# ---------------------------------------------
df = pd.read_csv("employee_attrition_preprocessed.csv")

# ---------------------------------------------
# 2. SET TARGET AND FEATURES
# ---------------------------------------------
target = "Attrition"

important_features = [
    "OverTime",
    "MonthlyIncome",
    "JobSatisfaction",
    "YearsAtCompany",
    "TotalWorkingYears",
    "YearsInCurrentRole",
    "JobLevel",
    "Age",
    "Gender",
    "HourlyRate",
    "PerformanceRating",
    "TrainingTimesLastYear",
    "NumCompaniesWorked",
    "JobInvolvement",
    "Department",
    "Education",
    "EducationField",
    "JobRole"
]

X = df[important_features]
y = df[target]

# ---------------------------------------------
# 3. TRAIN / TEST SPLIT
# ---------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

# ---------------------------------------------
# 4. RANDOM FOREST MODEL (simple)
# ---------------------------------------------
model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# TRAIN MODEL
model.fit(X_train, y_train)

# ---------------------------------------------
# 5. PREDICTIONS
# ---------------------------------------------
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

# ---------------------------------------------
# 6. METRICS
# ---------------------------------------------
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print("\n========== RANDOM FOREST (model.fit) RESULTS ==========\n")
print(f"Accuracy       : {acc:.4f}")
print(f"Precision      : {prec:.4f}")
print(f"Recall         : {rec:.4f}")
print(f"F1 Score       : {f1:.4f}")
print(f"AUC-ROC Score  : {auc:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, digits=4))

# ---------------------------------------------
# 7. ROC CURVE
# ---------------------------------------------
fpr, tpr, _ = roc_curve(y_test, y_proba)
plt.figure(figsize=(6,5))
plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
plt.plot([0,1], [0,1],"k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Random Forest (model.fit)")
plt.legend()
plt.show()

# ---------------------------------------------
# 8. FEATURE IMPORTANCE
# ---------------------------------------------
importances = model.feature_importances_
feat_imp = pd.Series(importances, index=important_features).sort_values(ascending=False)

plt.figure(figsize=(8,6))
feat_imp.plot.barh()
plt.gca().invert_yaxis()
plt.title("Feature Importances - Random Forest")
plt.xlabel("Importance Score")
plt.tight_layout()
plt.show()
```

```python
# full_xgb_or_rf_no_errors.py (paste into one notebook cell and run)
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt
from time import time

from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve, precision_recall_curve
)
from sklearn.ensemble import RandomForestClassifier

# Try XGBoost, otherwise fallback to RandomForest
USE_XGBOOST = True
try:
    from xgboost import XGBClassifier
except Exception:
    USE_XGBOOST = False
    print("xgboost not available — will use RandomForest fallback. To use XGBoost run: pip install xgboost")

# -------------------------
# 1) Load data
# -------------------------
df = pd.read_csv("employee_attrition_preprocessed.csv")
target = "Attrition"
if target not in df.columns:
    raise ValueError("Target 'Attrition' not found in CSV.")

# -------------------------
# 2) Feature engineering (high-impact)
# -------------------------
df = df.copy()

def safe_div(a,b):
    return a / (b + 1e-6)

# Base chosen features (only keep existing)
base_feats = [
    "OverTime", "MonthlyIncome", "JobSatisfaction", "YearsAtCompany",
    "TotalWorkingYears", "JobLevel", "Age", "EnvironmentSatisfaction", "JobRole"
]
base_feats = [c for c in base_feats if c in df.columns]

# Add engineered features if inputs exist
if 'YearsAtCompany' in df.columns:
    df['tenure_bucket'] = pd.cut(df['YearsAtCompany'], [-1,1,3,5,10,100],
                                labels=['0-1','2-3','4-5','6-10','10+'])
if {'MonthlyIncome','JobLevel'}.issubset(df.columns):
    df['income_ratio'] = safe_div(df['MonthlyIncome'], df['JobLevel']+1)
if {'OverTime','JobSatisfaction'}.issubset(df.columns):
    df['burnout_score'] = df['OverTime'] * (5 - df['JobSatisfaction'])
if {'YearsAtCompany','TotalWorkingYears'}.issubset(df.columns):
    df['loyalty_ratio'] = safe_div(df['YearsAtCompany'], df['TotalWorkingYears'])
if 'Age' in df.columns:
    df['age_group'] = pd.cut(df['Age'], [17,25,35,45,60], labels=['18-25','26-35','36-45','46+'])
if {'TotalWorkingYears','YearsAtCompany'}.issubset(df.columns):
    df['experience_gap'] = df['TotalWorkingYears'] - df['YearsAtCompany']
if {'MonthlyIncome','TotalWorkingYears'}.issubset(df.columns):
    df['salary_per_experience'] = safe_div(df['MonthlyIncome'], df['TotalWorkingYears'])
if {'YearsAtCompany','TotalWorkingYears'}.issubset(df.columns):
    df['role_stagnation'] = safe_div(df['YearsAtCompany'], df['TotalWorkingYears'])

engineered = [c for c in ['tenure_bucket','income_ratio','burnout_score','loyalty_ratio',
                          'age_group','experience_gap','salary_per_experience','role_stagnation'] if c in df.columns]

features = base_feats + engineered
print("Features used:", features)

X = df[features].copy()
y = df[target].copy()

# -------------------------
# 3) Train / test split
# -------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=42
)

# -------------------------
# 4) Robust numeric / categorical detection
# -------------------------
num_cols = X_train.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X_train.select_dtypes(include=['object','category']).columns.tolist()

print("Numeric cols:", num_cols)
print("Categorical cols:", cat_cols)

# Safe OneHotEncoder fallback for old/new sklearn
try:
    ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    ohe = OneHotEncoder(handle_unknown='ignore', sparse=False)

preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', ohe, cat_cols)
], remainder='drop')

# Fit/transform
X_train_proc = preprocessor.fit_transform(X_train)
X_test_proc  = preprocessor.transform(X_test)

# Save preprocessor
joblib.dump(preprocessor, "preprocessor_saved.joblib")
print("Saved preprocessor -> preprocessor_saved.joblib")

# -------------------------
# 5) Choose algorithm and hyperparam search
# -------------------------
cv = StratifiedKFold(n_splits=4, shuffle=True, random_state=42)

if USE_XGBOOST:
    # compute scale_pos_weight for imbalance
    pos = int(np.sum(y_train==1))
    neg = int(np.sum(y_train==0))
    scale_pos_weight = max(1.0, neg / max(1, pos))
    print(f"XGBoost scale_pos_weight: {scale_pos_weight:.3f} (pos:{pos} neg:{neg})")

    estimator = XGBClassifier(objective='binary:logistic', use_label_encoder=False, eval_metric='auc',
                              random_state=42, n_jobs=-1)
    estimator.set_params(scale_pos_weight=scale_pos_weight)

    param_dist = {
        'n_estimators': [100,200,400],
        'max_depth': [3,4,5,6],
        'learning_rate': [0.01, 0.03, 0.05, 0.1],
        'subsample': [0.6,0.8,1.0],
        'colsample_bytree': [0.6,0.8,1.0],
        'gamma': [0,0.1,0.2],
        'reg_alpha': [0, 0.01, 0.1],
        'reg_lambda': [1.0, 2.0, 5.0]
    }
    n_iter_search = 25
else:
    estimator = RandomForestClassifier(class_weight='balanced', random_state=42, n_jobs=-1)
    param_dist = {
        'n_estimators': [200,400,800],
        'max_depth': [6,10,16,None],
        'min_samples_leaf': [1,2,4],
        'min_samples_split': [2,4,8],
        'max_features': ['sqrt', 0.5, 0.75]
    }
    n_iter_search = 25

rnd = RandomizedSearchCV(
    estimator=estimator,
    param_distributions=param_dist,
    n_iter=n_iter_search,
    scoring='roc_auc',
    cv=cv,
    verbose=2,
    random_state=42,
    n_jobs=-1
)

# -------------------------
# 6) Run RandomizedSearchCV
# -------------------------
t0 = time()
print("Running RandomizedSearchCV (this may take a few minutes)...")
rnd.fit(X_train_proc, y_train)
print(f"RandomizedSearchCV finished in {time()-t0:.1f}s")
print("Best params:", rnd.best_params_)
print("Best CV AUC:", rnd.best_score_)

best_model = rnd.best_estimator_

# Final fit on all training data (no early stopping to avoid version issues)
best_model.set_params(verbosity=0) if USE_XGBOOST else None
best_model.fit(X_train_proc, y_train)

# -------------------------
# 7) Evaluate final model on holdout test
# -------------------------
y_proba = best_model.predict_proba(X_test_proc)[:,1] if hasattr(best_model, "predict_proba") else best_model.decision_function(X_test_proc)
y_pred = (y_proba >= 0.5).astype(int)

acc  = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, zero_division=0)
rec  = recall_score(y_test, y_pred)
f1s = f1_score(y_test, y_pred)
auc  = roc_auc_score(y_test, y_proba)

print("\n========== FINAL EVALUATION ==========\n")
print(f"Model used      : {'XGBoost' if USE_XGBOOST else 'RandomForest'}")
print(f"Accuracy        : {acc:.4f}")
print(f"Precision       : {prec:.4f}")
print(f"Recall          : {rec:.4f}")
print(f"F1 Score        : {f1s:.4f}")
print(f"AUC-ROC         : {auc:.4f}")
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, digits=4))

# -------------------------
# 8) Threshold tuning (maximize F1)
# -------------------------
precisions, recalls, thresholds = precision_recall_curve(y_test, y_proba)
f1_scores = 2*(precisions*recalls)/(precisions+recalls+1e-12)
best_idx = np.nanargmax(f1_scores)
best_threshold = thresholds[best_idx] if best_idx < len(thresholds) else 0.5
print(f"Best threshold by F1 on test = {best_threshold:.3f}, F1 = {f1_scores[best_idx]:.4f}")

y_pred_best = (y_proba >= best_threshold).astype(int)
print("\nMetrics after threshold tuning:")
print("Accuracy :", accuracy_score(y_test, y_pred_best))
print("Precision:", precision_score(y_test, y_pred_best, zero_division=0))
print("Recall   :", recall_score(y_test, y_pred_best))
print("F1       :", f1_score(y_test, y_pred_best))
print("Confusion:\n", confusion_matrix(y_test, y_pred_best))

# -------------------------
# 9) Feature importance mapping & plot
# -------------------------
# Build feature names: numeric columns first then OHE columns
feature_names = num_cols.copy()
try:
    cat_ohe = preprocessor.named_transformers_['cat']
    if hasattr(cat_ohe, "get_feature_names_out"):
        ohe_names = list(cat_ohe.get_feature_names_out(cat_cols))
    else:
        ohe_names = []
        for c in cat_cols:
            vals = X_train[c].astype(str).unique().tolist()[:50]
            for v in vals:
                ohe_names.append(f"{c}__{v}")
    feature_names += ohe_names
except Exception:
    feature_names = [f"f{i}" for i in range(X_train_proc.shape[1])]

# Extract importances
if USE_XGBOOST:
    importances = best_model.feature_importances_
else:
    importances = best_model.feature_importances_

imp_series = pd.Series(importances, index=feature_names).sort_values(ascending=False)
print("\nTop features:")
print(imp_series.head(20))

plt.figure(figsize=(10,6))
imp_series.head(20).plot.barh()
plt.gca().invert_yaxis()
plt.title("Top Feature Importances")
plt.tight_layout()
plt.show()

# -------------------------
# 10) Save model & preprocessor & predictions
# -------------------------
joblib.dump(best_model, "best_model.joblib")
joblib.dump(preprocessor, "preprocessor.joblib")
res = X_test.copy()
res['true'] = y_test.values
res['proba'] = y_proba
res['pred_0.5'] = y_pred
res['pred_best_thresh'] = y_pred_best
res.to_csv("model_test_predictions.csv", index=False)
print("Saved artifacts: best_model.joblib, preprocessor.joblib, model_test_predictions.csv")
```

```python
# knn_model
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import joblib
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report, roc_curve
)

# 1. LOAD DATA

df = pd.read_csv("employee_attrition_preprocessed.csv")
target = "Attrition"

# Convert Yes/No to 1/0 if needed
if df[target].dtype == "object":
    df[target] = df[target].map({"Yes": 1, "No": 0})

X = df.drop(columns=[target])
y = df[target]


# 2. TRAIN TEST SPLIT

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)


# 3. PREPROCESSING

num_cols = X_train_raw.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X_train_raw.select_dtypes(include=["object", "category"]).columns.tolist()

try:
    ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
except:
    ohe = OneHotEncoder(handle_unknown="ignore", sparse=False)

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), num_cols),
    ("cat", ohe, cat_cols)
])

X_train = preprocessor.fit_transform(X_train_raw)
X_test  = preprocessor.transform(X_test_raw)


# 4. GRID SEARCH FOR KNN

param_grid = {
    "n_neighbors": [3,5,7,9,11,15],
    "weights": ["uniform", "distance"],
    "p": [1,2]
}

knn = KNeighborsClassifier()
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

grid = GridSearchCV(
    knn,
    param_grid,
    scoring="accuracy",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid.fit(X_train, y_train)
best_knn = grid.best_estimator_

print("Best Parameters:", grid.best_params_)


# 5. EVALUATION

y_pred = best_knn.predict(X_test)
y_proba = best_knn.predict_proba(X_test)[:,1]

print("\n========== KNN RESULTS ==========\n")
print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("AUC      :", roc_auc_score(y_test, y_proba))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# 6. ROC CURVE

fpr, tpr, _ = roc_curve(y_test, y_proba)

plt.figure()
plt.plot(fpr, tpr, label="KNN ROC")
plt.plot([0,1], [0,1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - KNN")
plt.legend()
plt.show()

# 7. SAVE MODEL + PREPROCESSOR

joblib.dump(best_knn, "best_knn_model.joblib")
joblib.dump(preprocessor, "knn_preprocessor.joblib")

print("\n✅ Model and Preprocessor Saved Successfully")
```
